## Step 2 - Are the packages installed?

Each line tries to load (import) one package and prints its version.

```python
import importlib

packages = {
    "numpy": "numpy",
    "pandas": "pandas",
    "matplotlib": "matplotlib",
    "scikit-learn": "sklearn",
    "nltk": "nltk",
    "opencv-python": "cv2",
    "google-genai": "google.genai",
    "python-dotenv": "dotenv",
    "pydantic": "pydantic",
}

missing = []
for install_name, import_name in packages.items():
    try:
        module = importlib.import_module(import_name)
        version = getattr(module, "__version__", "installed")
        print(f"OK       {install_name:15} {version}")
    except ImportError:
        print(f"MISSING  {install_name}")
        missing.append(install_name)

if missing:
    print()
    print("Some packages are missing. In the terminal (with .venv active) run:")
    print("pip install -r requirements.txt")
```

## Step 3 - Can we draw a chart?

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4]
hours_of_learning = [6, 6, 6, 6]

plt.plot(days, hours_of_learning, marker="o")
plt.title("Hours of learning in this course")
plt.xlabel("Day")
plt.ylabel("Hours")
plt.show()
```
