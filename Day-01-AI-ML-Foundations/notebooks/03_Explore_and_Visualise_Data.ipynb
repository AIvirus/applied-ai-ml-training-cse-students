# Notebook 03 - Explore and Visualise Data (EDA)

Goes with **Lesson 05**. EDA = Exploratory Data Analysis = getting to know your data **before** building a model.

We follow the 8-step EDA checklist from the lesson.

```python
from pathlib import Path

# The data files are in the "data" folder next to "notebooks".
# In Google Colab, upload the CSV file and it will be found in the current folder.
DATA_DIR = Path("../data") if Path("../data").exists() else Path(".")
print("Reading data from:", DATA_DIR.resolve())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(DATA_DIR / "student_performance.csv")
```

## Step 1 - How big is it?

```python
print("Rows (students):", df.shape[0])
print("Columns:        ", df.shape[1])
```

## Step 2 - What do the first rows look like?

```python
df.head()
```

## Step 3 - Column types and missing values

- `int64` = whole numbers, `float64` = decimal numbers, `object` or `str` = text.
- **Non-Null Count** tells how many cells are filled. 300 everywhere means nothing is missing.

```python
df.info()
```

## Step 4 - Basic statistics for every number column

`25%`, `50%` and `75%` are **quartiles**: 25 percent of students are below the `25%` value. The `50%` value is the median.

```python
df.describe().round(1)
```

## Step 5 - How many of each category?

```python
for column in ["result", "city_type", "internet_access", "extra_classes"]:
    print(df[column].value_counts())
    print()
```

## Step 6 - Compare groups with groupby

```python
print("Average final score by extra classes:")
print(df.groupby("extra_classes")["final_score"].mean().round(1))
print()
print("Average final score by city type:")
print(df.groupby("city_type")["final_score"].mean().round(1))
print()
print("Average habits of students who passed and failed:")
df.groupby("result")[["hours_studied", "attendance_percent", "sleep_hours", "previous_score"]].mean().round(1)
```

## Step 7 - Which columns move together? (correlation)

```python
corr = df.corr(numeric_only=True).round(2)
corr
```

## Step 8 - Charts

### Histogram: how are the final scores spread?

```python
plt.figure(figsize=(8, 4))
plt.hist(df["final_score"], bins=15, edgecolor="black")
plt.axvline(50, color="red", linestyle="--", label="pass mark (50)")
plt.title("How final scores are spread")
plt.xlabel("Final score")
plt.ylabel("Number of students")
plt.legend()
plt.show()
```

### Bar chart: compare categories

```python
average_by_city = df.groupby("city_type")["final_score"].mean().sort_values()

plt.figure(figsize=(6, 4))
plt.bar(average_by_city.index, average_by_city.values)
plt.title("Average final score by city type")
plt.xlabel("City type")
plt.ylabel("Average final score")
plt.show()
```

### Scatter plot: relationship between two numbers

```python
passed = df[df["result"] == "pass"]
failed = df[df["result"] == "fail"]

plt.figure(figsize=(8, 5))
plt.scatter(passed["hours_studied"], passed["final_score"], label="pass", alpha=0.6)
plt.scatter(failed["hours_studied"], failed["final_score"], label="fail", alpha=0.6)
plt.title("Hours studied vs final score")
plt.xlabel("Hours studied per day")
plt.ylabel("Final score")
plt.legend()
plt.show()
```

### Box plot: spread and outliers

The box shows the middle 50 percent of students. The line inside is the median. Dots outside the whiskers are possible outliers.

```python
groups = [df[df["extra_classes"] == "no"]["final_score"], df[df["extra_classes"] == "yes"]["final_score"]]

plt.figure(figsize=(6, 4))
plt.boxplot(groups)
plt.xticks([1, 2], ["no extra classes", "extra classes"])
plt.title("Final score with and without extra classes")
plt.ylabel("Final score")
plt.show()
```

### Heatmap: the correlation table as colours

```python
plt.figure(figsize=(7, 5))
plt.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="correlation")
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45, ha="right")
plt.yticks(range(len(corr.columns)), corr.columns)
for i in range(len(corr)):
    for j in range(len(corr)):
        plt.text(j, i, corr.iloc[i, j], ha="center", va="center", fontsize=8)
plt.title("Correlation heatmap")
plt.show()
```
